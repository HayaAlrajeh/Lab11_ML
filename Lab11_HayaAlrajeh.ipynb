# Lab 11 : Credit Card Customer Segmentation using K-Means Clustering

### Import libraries 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA
```

### Load the dataset and display first rows

```python
df = pd.read_csv('CC_GENERAL.csv')
df.head()
```

### Check Dataset Shape

```python
df.shape
```

### Display Dataset information

```python
df.info()
```

### Statistical Summary

```python
df.describe()
```

### Handle Missing Values

```python
df = df.drop('CUST_ID', axis=1)

df.isnull().sum()
```

```python
df.fillna(df.mean(), inplace=True)

df.isnull().sum()
     
```

### Data Visualization

```python
df.hist(figsize=(20, 20))
plt.show()
```

### Correlation Analysis

```python
plt.figure(figsize=(15, 10))
sns.heatmap(df.corr(), cmap='coolwarm')
plt.show()
```

### Balance vs Purchases

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(x=df['BALANCE'], y=df['PURCHASES'])
plt.title('Balance vs Purchases')
plt.show()
     
```

### Balance vs Cash Advance

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(x=df['BALANCE'], y=df['CASH_ADVANCE'])
plt.title('Balance vs Cash Advance')
plt.show()
     
```

### Find Optimal Number

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)
```

```python
inertia_values = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertia_values.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), inertia_values, marker='o')
plt.xlabel('Number of Clusters')
plt.ylabel('Inertia')
plt.title('Elbow Method')
plt.show()
```

### Evaluate Cluster Performance

```python
silhouette_scores = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    silhouette_scores.append(score)

plt.figure(figsize=(8, 5))
plt.plot(range(2, 11), silhouette_scores, marker='o')
plt.xlabel('Number of Clusters')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Scores')
plt.show()
```

```python
scores_df = pd.DataFrame({
    'K': range(2, 11),
    'Silhouette Score': silhouette_scores
})

scores_df
     
```

### Apply K-Means Clustering

```python
final_k = 4

kmeans = KMeans(n_clusters=final_k, random_state=42, n_init=10)
clusters = kmeans.fit_predict(X_scaled)

df['Cluster'] = clusters

df.head()
     
```

### Cluster Result

```python
df['Cluster'].value_counts()
```

```python
cluster_summary = df.groupby('Cluster').mean()
cluster_summary
```

### PCA

```python
pca = PCA(n_components=2)
pca_components = pca.fit_transform(X_scaled)

pca_df = pd.DataFrame(pca_components, columns=['PCA1', 'PCA2'])
pca_df['Cluster'] = clusters

pca_df.head()
```

```python
plt.figure(figsize=(10, 7))
sns.scatterplot(
    x='PCA1',
    y='PCA2',
    hue='Cluster',
    data=pca_df,
    palette='Set2'
)
plt.title('Customer Segments using PCA')
plt.show()
```
